```python
import pandas as pd
```

```python
df = pd.read_csv('BankChurners.csv')
```

```python
df2 = df[['Customer_Age', 'Attrition_Flag']]
df2
```

```python
import matplotlib.pyplot as plt

# Group the age into 10s and create a new column 'Age Group'
df2.loc[:, 'Age Group'] = (df2['Customer_Age'] // 10) * 10

# Group the data by 'Age Group' and 'Attrition_Flag', and count the number of records in each group
result = df2.groupby(['Age Group', 'Attrition_Flag']).size().reset_index(name='Count')

# Pivot the data to create a stacked bar chart
pivot = result.pivot(index='Age Group', columns='Attrition_Flag', values='Count')

# Plot the stacked bar chart
pivot.plot(kind='bar', stacked=True)

# Add labels and title
plt.xlabel('Age Group')
plt.ylabel('Count')
plt.title('Attrition by Age Group')

# Show the plot
plt.show()
```
